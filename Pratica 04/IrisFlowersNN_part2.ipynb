# Importando bibliotecas

```python
import pandas as pd
import numpy as np

import tensorflow as tf
```

# Carregando e Preparando o DataSet

```python
from sklearn.datasets import load_iris
# Carregando o dataset Iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target).map({0: 'Setosa', 1: 'Versicolor', 2: 'Virginica'})
```

```python
df['species']=y
```

```python
df.head(10)
```

```python
df.info()
```

```python
df['species'].unique()
```

```python
df['species'] = df['species'].replace({'Setosa':0,
                                       'Versicolor':1,
                                       'Virginica':2}) # Adaptando os valores de string para numericos no estilo "sparse"
```

# Divisão treino-teste

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns=['species'])
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

# Padronizando os Dados

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler() # Instanciando padronizador de dados do sklearn
```

```python
scaler.fit(X_train) # Adequamos o padronizador aos dados de treino

X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

# Rede Neural

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(4,)),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

```python
model.compile(optimizer=tf.keras.optimizers.Adam(),
              loss=tf.keras.losses.SparseCategoricalCrossentropy(),
              metrics=['accuracy'])
```

# Treino

```python
BATCH_SIZE = 4
EPOCHS = 30
```

```python
model.fit(X_train, y_train, batch_size=BATCH_SIZE, epochs=EPOCHS)
```

# Testando modelo

```python
model.evaluate(X_test, y_test)
```

## Obtendo mais métricas

```python
y_pred = model.predict(X_test)
y_pred
```

```python
y_pred = [np.argmax(v) for v in y_pred]
y_pred
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred))
```
