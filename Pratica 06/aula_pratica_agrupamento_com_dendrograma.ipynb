```python
# Agrupamento hierárquico com dendrograma

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
import pandas as pd
```

```python
df = pd.read_csv('ruspini.csv')

X = df[['x', 'y']].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

plt.scatter(X[:, 0], X[:, 1])
plt.title("Conjunto de Dados Ruspini")
plt.show()
```

```python
# 3. Gerando a matriz de ligação
# Métodos comuns: 'ward', 'complete', 'average', 'single'
linked = linkage(X_scaled, method='ward')
linked
```

```python
# 4. Plotando o dendrograma
plt.figure(figsize=(10, 6))
dendrogram(linked)

plt.title("Dendrograma - Agrupamento Hierárquico")
plt.xlabel("Amostras")
plt.ylabel("Distância")
plt.grid(True)
plt.show()
```

```python
# Aplicando agrupamento hierárquico com 3 grupos

modelo = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

labels = modelo.fit_predict(X_scaled)

plt.figure(figsize=(7, 5))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, s=60)

plt.title("Agrupamento Hierárquico com 3 grupos")
plt.xlabel("Atributo 1")
plt.ylabel("Atributo 2")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
dendrogram(linked)

# Linha horizontal simulando um corte no dendrograma
plt.axhline(y=4, linestyle='--')

plt.title("Corte no Dendrograma")
plt.xlabel("Amostras")
plt.ylabel("Distância")
plt.grid(True)
plt.show()
```
